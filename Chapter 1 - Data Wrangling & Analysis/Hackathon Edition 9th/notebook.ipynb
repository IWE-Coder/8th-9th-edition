#1 Performance Analysis

```python
df_spot.set_index('Title')['Stream'].sort_values(ascending=False).head(10)
```

```python
df_spot.set_index('Title')['Views'].sort_values(ascending=False).head(10)
```

```python
df_spot.plot.scatter(
    x='Views',
    y='Stream',
    title='Spotify Streams vs YouTube Views',
    ylabel='Number of Spotify Streams',
    xlabel='Number of YouTube Views'
)
correlation = df_spot['Views'].corr(df_spot['Stream'])
print(correlation)
```

From comparing the songs between when they're sorted by Spotify Streams and by YouTube Views, we can see that aside from the song Shape of You by Ed Sheeran is the only common song at the Top 10. Also, when plotting a scatter plot between the two we can see that there seems to be very little correlation between them. When calculating the correlation coefficient, we get around 0.61, which isn't a high or low amount of correlation. So, there seems to be little relationship between getting many views on Youtube and getting many streams on Spotify.

Let's take a look at the relation between the amount of YouTube views and the number of Comments and Likes.

```python
df_spot.plot.scatter(
    x='Views',
    y='Comments',
    title='YouTube Views vs. YouTube Comments',
    ylabel='Number of YouTube Comments',
    xlabel='Number of YouTube Views'
)
correlation = df_spot['Views'].corr(df_spot['Comments'])
print(correlation)
```

```python
df_spot.plot.scatter(
    x='Views',
    y='Likes',
    title='YouTube Views vs. YouTube Likes',
    ylabel='Number of YouTube Likes',
    xlabel='Number of YouTube Views'
)
correlation = df_spot['Views'].corr(df_spot['Likes'])
print(correlation)
```

Interestingly, there is a very low correlation between the number of views on a Youtube MV and the number of comments, but a very high correlation between the number of views and the number of likes.
